# 🔤 Parte 2 · Variáveis e tipos de dados

Uma **variável** é como uma **célula com nome**: você guarda um valor e usa depois. Em vez de `B2`, a gente escolhe um nome que faz sentido (`preco`).

- **Texto** vai entre aspas: `"Cappuccino"`
- **Números** com casas decimais usam **ponto**: `12.90` (não `12,90`)
- Os 4 tipos mais comuns: texto (`str`), inteiro (`int`), decimal (`float`) e verdadeiro/falso (`bool`).

````python
produto = "Cappuccino"        # texto
preco = 12.90                 # decimal
quantidade = 3                # inteiro
e_delivery = True             # verdadeiro/falso

total = round(preco * quantidade, 2)      # round(..., 2) = arredondar para 2 casas
print("Pedido:", quantidade, "x", produto, "= R$", total)

print(type(produto), type(preco), type(quantidade), type(e_delivery))
````

A aba de produtos da planilha do café traz o preço como texto: `"R$ 12,90"`. Vamos precisar de uma função que **limpe** isso. Ela:

### 7.2 A fórmula arrastada: conferir o `Total (R$)` do Excel
Na planilha, a coluna `Total (R$)` veio **pronta** (valores, não fórmulas). Será que está certa? Vamos **recalcular** e comparar, o equivalente a criar uma coluna com `=E2*F2` e arrastar.

Em Python, você faz a conta com a **coluna inteira de uma vez**, sem arrastar nada.

````python
# ✏️ COMPLETE: troque os ___ pelo código certo
vendas["total_calc"] = vendas["quantidade"] * vendas["___"]      # quantidade x preço, linha a linha

maior_diferenca = (vendas["total_calc"] - vendas["total"]).abs().max()
print("Maior diferença entre o Excel e o Python: R$", round(maior_diferenca, 2))

conferir(maior_diferenca < 0.01, "O total do Excel bate com o cálculo do Python.", "a coluna deve ser quantidade vezes preco.")
````

### 7.3 SOMA, CONT.VALORES e o ticket médio
- `SOMA` → `.sum()`
- Quantos pedidos **diferentes**? (um pedido pode ter vários itens) → `.nunique()`
- Ticket médio = faturamento ÷ pedidos

````python
# ✏️ COMPLETE: troque os ___ pelo código certo
faturamento = vendas["total"].___()          # SOMA
pedidos = vendas["pedido"].___()             # quantos pedidos diferentes? (nunique)
ticket_medio = faturamento / pedidos

print(f"Faturamento: R$ {faturamento:,.2f} | Pedidos: {pedidos} | Ticket médio: R$ {ticket_medio:,.2f}")

conferir(round(faturamento, 2) == 46112.1 and pedidos == 2219, "Faturamento de setembro: R$ 46.112,10.", "o faturamento deve ser 46112.1 (soma de total) e os pedidos 2219 (nunique da coluna pedido).")
````

### 7.4 SOMASE: faturamento por canal
`=SOMASE(canal; "iFood"; total)` soma o total **para cada canal**. Em pandas isso é **agrupar** (`groupby`) e somar.

```
tabela.groupby("coluna_para_agrupar")["coluna_para_somar"].sum()
```

````python
# ✏️ COMPLETE: troque os ___ pelo código certo
fat_canal = vendas.groupby("___")["total"].sum()        # agrupa por canal e soma o total
fat_canal = fat_canal.round(2)
fat_canal
````

Na tabela dinâmica do Excel você põe o **produto nas linhas**, **quantidade nos valores** e **ordena do maior para o menor**. Em pandas: agrupar → somar → ordenar → pegar os 5 primeiros.

🧠 **Pense:** para ordenar do **maior para o menor**, o `ascending` (crescente) deve ser `True` ou `False`?

### 7.6 PROCV: trazer o nome do produto
As vendas só têm o código (`P03`). O nome e a categoria estão na aba **Produtos**, o clássico `PROCV`. Em pandas: `merge`.

Mas atenção: nessa aba, o preço e o custo vêm como **texto** (`"R$ 12,90"`). Hora de usar a função `para_numero` que você criou na Parte 5!

````python
# ✏️ COMPLETE: troque os ___ pelo código certo
produtos = ler_excel(ARQUIVO, sheet_name="Produtos")

produtos["preco_atual"] = produtos["preco_atual"].apply(___)         # a função da Parte 5
produtos["custo_unitario"] = produtos["custo_unitario"].apply(para_numero)

vendas_prod = vendas.merge(
    produtos[["id_produto", "nome_produto", "categoria", "custo_unitario"]],
    on="___",          # coluna que existe nas duas tabelas
    how="left",
)

conferir(len(vendas_prod) == len(vendas) and vendas_prod["nome_produto"].notna().all() and produtos["preco_atual"].dtype == "float64", "PROCV feito! Cada venda agora tem nome e categoria.", "use apply(para_numero) e faça o merge pela coluna id_produto.")
vendas_prod.head()
````

Com o custo na mão, dá para calcular o **lucro** de cada linha (receita menos custo) e ver qual produto dá mais lucro, algo que o dono do café adora:

````python
vendas_prod["lucro"] = vendas_prod["total"] - vendas_prod["quantidade"] * vendas_prod["custo_unitario"]

lucro_produto = vendas_prod.groupby("nome_produto")["lucro"].sum().sort_values(ascending=False).round(2)
lucro_produto.head(5)
````

### 7.7 Comparar com as metas
A aba **Metas** tem a meta de faturamento por categoria, mas também tem um título antes da tabela. Mesmo truque do `header`.

🧠 **Pense:** o cabeçalho está na **linha 3** do Excel. Qual o número (contando do zero)?

````python
# ✏️ COMPLETE: troque os ___ pelo código certo
metas = ler_excel(ARQUIVO, sheet_name="Metas", header=___)
metas.columns = ["categoria", "meta"]

fat_cat = vendas_prod.groupby("categoria", as_index=False)["total"].sum().rename(columns={"total": "faturamento"})
fat_cat = fat_cat.merge(metas, on="categoria")

fat_cat["diferenca"] = fat_cat["faturamento"] - fat_cat["___"]       # quanto passou (ou faltou) da meta
fat_cat["situacao"] = fat_cat.apply(lambda linha: situacao_meta(linha["faturamento"], linha["meta"]), axis=1)

fat_cat.round(2)
````

> 💡 O `apply(..., axis=1)` aplica a função **linha por linha**: é exatamente o "arrastar a fórmula para baixo" do Excel, usando a função `situacao_meta` que você viu na Parte 5. Não precisa decorar. Quando precisar, peça à IA.

````python
# ✅ Conferência 7.7
bateu = set(fat_cat.loc[fat_cat["situacao"].str.contains("bateu"), "categoria"])
conferir(bateu == {"Bebida fria", "Doce"}, "Bebida fria e Doce bateram a meta; Bebida quente e Salgado ficaram abaixo.", "só Bebida fria e Doce deveriam bater a meta. Confira o header das metas e a coluna da diferença.")
````

Esta célula faz **todo o caminho da Parte 6 de uma vez** (cabeçalho, rodapé, nomes e tipos) e deixa `vendas`, `para_numero` e `situacao_meta` prontos, para você seguir direto para a **Parte 7**.

````python
# 🆘 Apêndice B: atalho com tudo da Parte 6 pronto
def para_numero(texto):
    return float(str(texto).replace("R$", "").strip().replace(",", "."))

def situacao_meta(valor, meta):
    return "✅ bateu a meta" if valor >= meta else "⚠️ abaixo da meta"

vendas = ler_excel(ARQUIVO, sheet_name="Vendas Set", header=3).dropna(subset=["Pedido"]).reset_index(drop=True)
vendas = vendas.rename(columns={
    "Data": "data", "Hora": "hora", "Pedido": "pedido", "Cód. Produto": "id_produto",
    "Qtde.": "quantidade", "Preço Unit. (R$)": "preco", "Canal": "canal",
    "Pagamento": "pagamento", "Total (R$)": "total",
})
vendas["data"] = pd.to_datetime(vendas["data"])
vendas["hora"] = vendas["hora"].astype(int)
vendas["quantidade"] = vendas["quantidade"].astype(int)

print(f"Tabela vendas pronta: {len(vendas):,} linhas ✅".replace(",", "."))
````
